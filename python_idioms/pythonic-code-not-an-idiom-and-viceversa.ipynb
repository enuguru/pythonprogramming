# python idiom

```python
my_dict = {"focus":100,"no_focus":0}
```

```python
try:
    value = my_dict['focus']
except KeyError:
    value = None
print(value)
```

```python
value = my_dict.get('focus', None)
print(value)
value = my_dict.get('yes', None)
print(value)
```

# pythonic code

```python
value = my_dict.get('key', None)
print(value)
```

# pythonic code but not an idiom

```python
x = 5
y = 10
max_value = x if x > y else y
print(max_value)
```
