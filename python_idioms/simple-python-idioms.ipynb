```python
numbers = [1, 2, 3, 4, 5]

# Check if any number is greater than 4
print(any(x > 4 for x in numbers))  # Output: True

# Check if all numbers are positive
print(all(x > 0 for x in numbers))  # Output: True
```

```python
# Unpacking a tuple
coordinates = (10, 20)
x, y = coordinates
print(x, y)  # Output: 10, 20

# Unpacking with star expression
a, *b, c = [1, 2, 3, 4, 5]
print(a, b, c)  # Output: 1 [2, 3, 4] 5
```

```python
# Using enumerate() to Loop with Indices
items = ['apple', 'banana', 'cherry']

for index, item in enumerate(items):
    print(f"{index}: {item}")
# Output:
# 0: apple
# 1: banana
# 2: cherry
```

```python
# Zipping Lists Together
# The zip() function allows you to iterate over multiple lists in parallel
```

```python
names = ["Alice", "Bob", "Charlie"]
ages = [25, 30, 35]

for name, age in zip(names, ages):
    print(f"{name} is {age} years old")
# Output:
# Alice is 25 years old
# Bob is 30 years old
# Charlie is 35 years old
```

```python
# Create a dictionary of squares
squares = {x: x**2 for x in range(5)}
print(squares)  # Output: {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}
```

```python
# Using get() with Dictionaries
person = {"name": "Alice", "age": 25}

# Safely get a key with a default value
city = person.get("city", "Unknown")
print(city)  # Output: Unknown
```

```python
numbers = [1, 2, 2, 3, 4, 4, 5]
unique_numbers = list(set(numbers))
print(unique_numbers)  # Output: [1, 2, 3, 4, 5]
```

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 3, "c": 4}

merged_dict = dict1 | dict2
print(merged_dict)  # Output: {'a': 1, 'b': 3, 'c': 4}
```

```python
# Using try-except-else-finally for Exception Handling
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero")
    else:
        print(f"Result: {result}")
    finally:
        print("Execution completed")

divide(10, 2)
divide(10, 0)
```

```python
# Using with Statement for Managing Resources
# The with statement is used to manage resources, 
# like file streams, without explicitly closing them
```

```python
# Automatically closes the file when the block is exited
with open('example.txt', 'w') as file:
    file.write("Hello, World!")
```

```python
# Default Values for Function Parameters
def greet(name="Guest"):
    print(f"Hello, {name}")

greet()  # Output: Hello, Guest
greet("Alice")  # Output: Hello, Alice
```

```python
# Using is for Singleton Comparison
# Checking if a variable is None
x = None
if x is None:
    print("x is None")
```

```python
# Using in for Membership Testing
fruits = ["apple", "banana", "cherry"]

if "banana" in fruits:
    print("Banana is in the list")
```

```python
# Using sorted() with a Key Function
students = [
    {"name": "Alice", "age": 25},
    {"name": "Bob", "age": 20},
    {"name": "Charlie", "age": 23},
]

# Sort students by age
sorted_students = sorted(students, key=lambda s: s["age"])
print(sorted_students)
```

```python
# Using _*_ for Unused Variables
# Looping through a range without using the variable
for _ in range(5):
    print("Hello")
```

```python
# Using collections.defaultdict for Default Values in Dictionaries
from collections import defaultdict

# Create a defaultdict with a default value of 0
counts = defaultdict(int)
counts["apple"] += 1
print(counts)  # Output: defaultdict(<class 'int'>, {'apple': 1})
```

```python
# Using itertools for Infinite Loops or Combinatorics
import itertools

# Infinite loop (press Ctrl+C to stop)
for x in itertools.count(10):
    if x > 15:
        break
    print(x)  # Output: 10, 11, 12, 13, 14, 15
```

```python
from functools import lru_cache

@lru_cache(maxsize=100)
def fibonacci(n):
    if n < 2:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(10))  # Output: 55
```

```python
def my_generator(n):
    for i in range(n):
        yield i ** 2

gen = my_generator(5)
for val in gen:
    print(val)  # Output: 0, 1, 4, 9, 16
```

```python
def my_function(*args, **kwargs):
    print("Positional arguments:", args)
    print("Keyword arguments:", kwargs)

my_function(1, 2, 3, name="Alice", age=30)
```
