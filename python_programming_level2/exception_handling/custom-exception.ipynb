```python
# Define a custom exception class
class NegativeNumberError(Exception):
    """Custom exception raised when a negative number is encountered."""
    def __init__(self, value):
        self.value = value
        super().__init__(f"NegativeNumberError: {value} is a negative number")

# Function that raises the custom exception
def check_positive_number(number):
    if number < 0:
        # Raise the custom exception when a negative number is found
        raise NegativeNumberError(number)
    else:
        return f"{number} is a positive number."

# Main code to demonstrate usage
try:
    number = int(input("Enter a number: "))
    result = check_positive_number(number)
    print(result)
except NegativeNumberError as e:
    # Catch and handle the custom exception
    print(e)
```

```python
class Exception(BaseException):
    """Common base class for all non-exit exceptions."""
    
    def __init__(self, *args):
        # Exception stores arguments in its args attribute
        self.args = args

    def __str__(self):
        # Converts the exception to a string
        if len(self.args) > 0:
            return str(self.args[0])
        else:
            return ""

    def __repr__(self):
        # Provides a developer-friendly string representation of the object
        return f"{self.__class__.__name__}{self.args}"
```

```python
try:
    # Raise an exception with a message
    raise Exception("Something went wrong!")
except Exception as e:
    print(str(e))  # Output: Something went wrong!
    print(repr(e))  # Output: Exception('Something went wrong!',)
```

```python
class BaseException:
    """The base class for all built-in exceptions."""
    
    def __init__(self, *args):
        # The arguments that were passed when the exception was raised
        self.args = args

    def __str__(self):
        # Convert the exception to a string for a user-friendly message
        if len(self.args) > 0:
            return str(self.args[0])
        else:
            return ""

    def __repr__(self):
        # Developer-friendly string representation of the exception object
        return f"{self.__class__.__name__}{self.args}"

    def __del__(self):
        """Destructor to clean up resources (if needed)"""
        pass

    def with_traceback(self, tb):
        """Sets a traceback for the exception."""
        self.__traceback__ = tb
        return self

    @property
    def __traceback__(self):
        """Retrieves the traceback associated with the exception."""
        return self._traceback

    @__traceback__.setter
    def __traceback__(self, tb):
        """Sets the traceback."""
        self._traceback = tb
```

```python
try:
    raise BaseException("A base exception occurred!")
except BaseException as e:
    print(repr(e))  # Output: BaseException('A base exception occurred!',)
```
