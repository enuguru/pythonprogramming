```python
"""
This is the module-level docstring.

This module demonstrates how to use docstrings for documenting classes and functions.
It contains a simple example of a `Calculator` class with various methods for arithmetic operations.
"""

class Calculator:
    """
    The Calculator class provides methods for basic arithmetic operations.
    
    Attributes:
        None.
    """
    
    def add(self, a, b):
        """
        Add two numbers and return the result.
        
        Args:
            a (float): The first number.
            b (float): The second number.
        
        Returns:
            float: The sum of the two numbers.
        
        Example:
            >>> calc = Calculator()
            >>> calc.add(3, 5)
            8
        """
        return a + b

    def subtract(self, a, b):
        """
        Subtract two numbers and return the result.
        
        Args:
            a (float): The first number.
            b (float): The second number.
        
        Returns:
            float: The result of subtracting b from a.
        
        Example:
            >>> calc = Calculator()
            >>> calc.subtract(10, 4)
            6
        """
        return a - b

    def multiply(self, a, b):
        """
        Multiply two numbers and return the result.
        
        Args:
            a (float): The first number.
            b (float): The second number.
        
        Returns:
            float: The product of the two numbers.
        
        Example:
            >>> calc = Calculator()
            >>> calc.multiply(2, 5)
            10
        """
        return a * b

    def divide(self, a, b):
        """
        Divide two numbers and return the result.
        
        Args:
            a (float): The first number (dividend).
            b (float): The second number (divisor).
        
        Returns:
            float: The quotient of the division.
        
        Raises:
            ZeroDivisionError: If the second number (divisor) is zero.
        
        Example:
            >>> calc = Calculator()
            >>> calc.divide(10, 2)
            5.0
        """
        if b == 0:
            raise ZeroDivisionError("Cannot divide by zero!")
        return a / b

if __name__ == "__main__":
    calc = Calculator()
    print(calc.add(3, 5))  # Output: 8
    print(calc.subtract(10, 4))  # Output: 6
    print(calc.multiply(2, 5))  # Output: 10
    print(calc.divide(10, 2))  # Output: 5.0
```
