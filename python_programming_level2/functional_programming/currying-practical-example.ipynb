```python
def calculate_total_price(discount):
    def with_tax(tax_rate):
        def with_delivery(delivery_charge):
            def total_price(item_price):
                # Apply discount, tax, and delivery charge to calculate the final price
                discounted_price = item_price - (item_price * discount)
                price_with_tax = discounted_price + (discounted_price * tax_rate)
                final_price = price_with_tax + delivery_charge
                return final_price
            return total_price
        return with_delivery
    return with_tax

# Create specialized versions using currying
# Scenario 1: Online sale (10% discount, 8% tax, $5 delivery charge)
online_sale = calculate_total_price(0.10)(0.08)(5.0)

# Scenario 2: In-store sale (15% discount, 5% tax, no delivery charge)
instore_sale = calculate_total_price(0.15)(0.05)(0.0)

# Calculate total price for an item costing $100 in both scenarios
online_total = online_sale(100)
instore_total = instore_sale(100)

print(f"Online Sale Total: ${online_total:.2f}")   # Output: $102.60
print(f"In-store Sale Total: ${instore_total:.2f}")  # Output: $91.25
```

```python
from functools import partial

def calculate_total_price(discount, tax_rate, delivery_charge, item_price):
    discounted_price = item_price - (item_price * discount)
    price_with_tax = discounted_price + (discounted_price * tax_rate)
    final_price = price_with_tax + delivery_charge
    return final_price

# Partially applying functions for different scenarios
online_sale = partial(calculate_total_price, 0.10, 0.08, 5.0)
instore_sale = partial(calculate_total_price, 0.15, 0.05, 0.0)

# Calculate total price for an item costing $100
online_total = online_sale(100)
instore_total = instore_sale(100)

print(f"Online Sale Total: ${online_total:.2f}")   # Output: $102.60
print(f"In-store Sale Total: ${instore_total:.2f}")  # Output: $91.25
```
