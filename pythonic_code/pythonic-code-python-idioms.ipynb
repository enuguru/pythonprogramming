# pythonic code

```python
result = []
for i in range(10):
    if i % 2 == 0:
        result.append(i)
print(result)
```

```python
result = [i for i in range(10) if i % 2 == 0]
print(result)
```

```python
x = 10
num_of_oranges = 10
print(x)
print(num_of_oranges)
```

```python
i = 0
for element in ['apple', 'banana', 'cherry']:
    print(i, element)
    i += 1

for i, element in enumerate(['apple', 'banana', 'cherry']):
    print(i, element)
```

```python
names = ['Alice', 'Bob', 'Charlie']
ages = [25, 30, 35]
for i in range(len(names)):
    print(f"{names[i]} is {ages[i]} years old.")

names = ['Alice', 'Bob', 'Charlie']
ages = [25, 30, 35]
for name, age in zip(names, ages):
    print(f"{name} is {age} years old.")
```

```python
person = {'name': 'John', 'age': 30}
if 'age' in person:
    age = person['age']
else:
    age = None

    
person = {'name': 'John', 'age': 30}
age = person.get('age')
```
